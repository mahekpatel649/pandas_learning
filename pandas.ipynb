```python
# to change data type of salary for median
# here salary data type is objrct
# so we have to convert it data type in float for fill other data type
df['salary']= df['salary'].str.replace(",","", regex=False)

# here all the value we got error that save here and program will continue
# bad value become NaN,not a crash
df['salary']= pd.to_numeric(df['salary'],errors = 'coerce')

# here salary null values are filled
df['salary']=df['salary'].fillna(df['salary'].median())
```

```python
# Grouping and Aggregation
# for this u can use groupby()
# it works in three stages:split the data into group,apply a calculation to each group separatly,then cambine the results back into one table
df.groupby("city")   # now city is one kind of separete object

```

```python
df.groupby("city")["age"].mean()  # you can get mean of age in separate city
```

```python
# to see max salary in some city
df.groupby("city")['salary'].sum()
```

```python
# there are other aggreations
# 1. sum() -total sum of all elements
# 2. mean() - average of elelment
# 3. median() - middle value
# 4. min() - miniumum value
# 5. max() - maximum value
# 6. count() - total lenght/size of axis// only count the non-null ones
# 7. std() - simple standard deviation
# 8. size() - count every row in the group

```

```python
# grouping muliple columns & aggregating multiple columns
df.groupby(["city","department"])["salary"].mean().round()
```

```python
# now lets make department as a main source
df.groupby(['department','city'])['salary'].mean().round()
```

```python
df.groupby("department")[["salary","performance_score"]].mean()  # multiple columns at once
```

```python
# multiple function at a time on multiple columns
df.groupby("department").agg({
    "salary":["mean","max"],
    "performance_score":"mean",
    "employee_id":"count"
}
)
```

```python
department_summary = df.groupby("department").agg(
    average_salary = ("salary","mean"),
    hightest_salary= ("salary","max"),
    employee_count =("employee_id","count")
)
```
